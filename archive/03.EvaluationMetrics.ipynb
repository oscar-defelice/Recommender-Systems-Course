# Evaluation Metrics in Recommender Systems

Recommender systems aim to deliver the most relevant items to users. But how do we know if a recommender system is actually performing well? Evaluation is crucial for understanding:

* ✅ How accurate the recommendations are.
* ✅ Whether the recommendations are diverse, novel, and serendipitous.
* ✅ Real-world user satisfaction (using A/B tests and online metrics).

There are two main approaches to evaluating recommender systems:

* 📊 Offline Evaluation: Uses historical data to test model performance without involving real users. 
* 📈 Online Evaluation: Involves real users and measures their interactions with the recommendations.

## Offline Evaluation

Offline evaluation is the first step in assessing the performance of a recommender system. It involves using historical data—typically split into training and test sets—to simulate recommendations and measure their accuracy against known user preferences. This type of evaluation is computationally inexpensive and allows for rapid experimentation with different algorithms and parameter settings. 

It helps data scientists fine-tune models before exposing them to real users. Common metrics used in offline evaluation include *RMSE* (Root Mean Squared Error), *MAE* (Mean Absolute Error), *Precision@K*, *Recall@K*, and *NDCG* (Normalized Discounted Cumulative Gain). However, offline evaluation has its limitations—it can only measure how well a system replicates historical data, not how it would perform in real-world scenarios where user behaviour is dynamic.

## Online Evaluation

Online evaluation involves deploying the recommender system in a real-world setting and measuring user interactions with the recommendations. 

This approach provides valuable insights into how users engage with the system and can help identify areas for improvement. Common metrics used in online evaluation include *click-through rate* (CTR), *conversion rate*, and *user satisfaction* (measured through surveys or A/B tests). However, online evaluation can be time-consuming and resource-intensive, and it may not always be feasible to deploy the system in a real-world setting.

## Offline vs Online Evaluation

| Offline Evaluation | Online Evaluation |
| --- | --- |
| Computationally inexpensive | Time-consuming and resource-intensive |
| Measures how well the system replicates historical data | Measures how users engage with the system |
| Can be used to fine-tune models before deployment | Can help identify areas for improvement in real-world scenarios |
| May not always capture real-world user behaviour | Provides valuable insights into user engagement and satisfaction |

## A/B Testing

A/B testing is a common approach used in online evaluation to compare the performance of two or more versions of a recommender system. In an A/B test, users are randomly assigned to one of the versions, and their interactions with the system are measured. The version that performs better according to the chosen metric is then selected for deployment.

In other words, A/B Testing is a controlled experiment widely used to evaluate the real-world impact of different versions of a recommender system. Unlike offline evaluation methods (like RMSE or NDCG), A/B testing measures actual user behaviour in response to changes in the recommendation algorithm.

### What Is A/B Testing?

A/B testing involves showing two (or more) versions of a recommender system to different user groups and measuring how these groups behave.

* Group A (Control Group) → Sees the existing system.
* Group B (Variant Group) → Sees the new or experimental system.

The goal is to determine whether the changes in the recommendation algorithm lead to improved user engagement or other key performance indicators (KPIs).

2. Choose the metrics to measure.

    **Running a T-Test on the Table**

A t-test is a statistical method used to determine if there is a significant difference between the means of two groups, which may be related in certain features. Here's a concise explanation of how you can run a t-test over the given table:

*Metrics*

| Metric       | Group A (Control) | Group B (Treatment) |
|--------------|-------------------|---------------------|
| CTR          | 0.05              | 0.055               |
| Conversion   | 0.02              | 0.023               |
| Time Spent   | 15 min            | 17 min              |
| Bounce Rate  | 0.10              | 0.09                |
| Diversity    | 0.70              | 0.75                |
| Churn Rate   | 0.03              | 0.025               |

*Steps to Run a T-Test*

1. **Identify the Metrics**: You have six metrics: CTR, Conversion, Time Spent, Bounce Rate, Diversity, and Churn Rate. Each metric has values for two groups: Group A (Control) and Group B (Treatment).

2. **Formulate Hypotheses**:
   - **Null Hypothesis $H_0$*: There is no significant difference between the means of Group A and Group B for a given metric.
   - **Alternative Hypothesis $H_1$**: There is a significant difference between the means of Group A and Group B for a given metric.

3. **Choose the Type of T-Test**:
   - **Paired T-Test**: Use this if the data in Group A and Group B are paired or matched.
   - **Independent T-Test**: Use this if the data in Group A and Group B are independent of each other.

4. **Check Assumptions**:
   - The data should be approximately normally distributed.
   - The variances of the two groups should be approximately equal (homogeneity of variances).

5. **Calculate the T-Statistic**:
   - For each metric, calculate the t-statistic using the formula:
     $$
     t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}
     $$
     where $(\bar{X}_1)$ and $(\bar{X}_2)$ are the means of the two groups, $(s_1^2)$ and $(s_2^2)$ are the variances, and $(n_1)$ and $(n_2)$ are the sample sizes.

6. **Determine the Degrees of Freedom**:
   - For an independent t-test, the degrees of freedom (df) can be approximated using the Welch-Satterthwaite equation. In practice, when the sample sizes are large enough, you can assume the t-statistics to be well approximated by a Gaussian function and use z-value to find the p-value.

7. **Find the P-Value**:
   - Use the t-statistic and degrees of freedom to find the p-value from the t-distribution table.

8. **Compare P-Value to Significance Level**:
   - If the p-value is less than the significance level (commonly $0.05$), reject the null hypothesis. This means there is a significant difference between the groups for that metric.

9. **Interpret the Results**:
   - For each metric, interpret whether the difference between Group A and Group B is statistically significant.

### Offline Analysis

Even if the online evaluation can be useful, it is not always possible to perform it. Therefore, we will perform the analysis offline. This is the case in model development where we need to perform the analysis on the data that is not available online.

Let's look in some detail at the metrics we commonly use in offline evaluation of recommender systems.

## 📊 **Detailed Presentation of Offline Evaluation Metrics for Recommender Systems**

**Offline evaluation** plays a crucial role during the **development and tuning** of recommender systems. It allows us to evaluate algorithms using **historical data** without deploying models into production or involving real users.  

The main goal is to assess **accuracy**, **relevance**, and other **key aspects** of the recommendations to ensure they meet desired performance standards before conducting **online tests** (like A/B testing).

---

### 📌 **1. Why Use Offline Evaluation?**  
✅ **Fast experimentation** without needing real-time user involvement.  
✅ **Low risk** – bad models don’t affect end-users.  
✅ **Objective comparison** of multiple algorithms.  
✅ **Helps fine-tune hyperparameters** before deployment.  

**⚠️ Limitation:** Offline metrics **cannot capture real user satisfaction** or behaviours (e.g., novelty preference, session time).

---

### 📊 **2. Key Offline Evaluation Metrics**

#### 🟢 **2.1 Accuracy-Based Metrics** *(Focus: Relevance of recommendations)*

These metrics evaluate how well the recommendations **match user preferences**.

---

#### **2.1.1 Precision@K**  
**Precision@K** measures the **proportion of relevant items** among the top-K recommended items.

$$
\text{Precision@K} = \frac{\text{number of relevant recommended items in top-K}}{K}
$$

- **High Precision@K** → Users are more likely to find something they like within the first K items.  
- **Best for systems prioritizing **“hit rate”** (e.g., e-commerce product suggestions).

**Example:**  
If 3 out of the top 5 recommended movies are relevant:

$$
\text{Precision@5} = \frac{3}{5} = 0.6
$$

---

#### **2.1.2 Recall@K**  
**Recall@K** focuses on how many of the total relevant items for a user are retrieved within the top-K recommendations.

$$
\text{Recall@K} = \frac{\text{number of relevant recommended items in top-K}}{\text{Total number of relevant items}}
$$

- **High Recall@K** → The system retrieves most relevant items for a user.  
- **Best for domains like medical recommendations or legal documents where missing relevant items is costly.**

**Example:**  
If the user has 8 relevant movies in total and the system retrieves 4 of them in the top 10:

$$
\text{Recall@10} = \frac{4}{8} = 0.5
$$

---

#### **2.1.3 F1-Score@K**  
**F1-Score** balances **Precision** and **Recall**. It is the **harmonic mean** of Precision@K and Recall@K.

$$
\text{F1@K} = 2 \times \frac{\text{Precision@K} \times \text{Recall@K}}{\text{Precision@K} + \text{Recall@K}}
$$

- Useful when both **precision and recall are important**.

---

#### **2.1.4 Mean Average Precision (MAP@K)**  
**MAP@K** takes into account **the ranking order** of relevant recommendations and calculates the **mean precision** at each point where a relevant item is retrieved.

$$
\text{MAP@K} = \frac{1}{N} \sum_{u=1}^{N} \frac{1}{|R_u|} \sum_{k=1}^{K} P@k \cdot rel(k)
$$

Where:  
- \( P@k \) = Precision at cut-off k  
- \( rel(k) \) = 1 if item at position k is relevant, else 0  
- \( N \) = Number of users  
- \( R_u \) = Relevant items for user u  

**MAP** rewards recommendations that **rank relevant items higher**.

---

#### **2.1.5 Normalised Discounted Cumulative Gain (NDCG@K)**  
**NDCG** evaluates the quality of the ranking by giving **higher importance to top-ranked relevant items**.

$$
\text{DCG@K} = \sum_{i=1}^{K} \frac{rel_i}{\log_2(i + 1)}
$$

$$
\text{NDCG@K} = \frac{\text{DCG@K}}{\text{IDCG@K}}
$$

Where:  
- \( rel_i \) = relevance of item at position i (binary or graded relevance)  
- \( IDCG@K \) = DCG of an ideal ranking  

**Why use NDCG?**  
- Penalises relevant items that are **ranked lower**.  
- Captures the fact that users are more likely to click on **top-ranked items**.

---

#### 🟡 **2.2 Error-Based Metrics** *(Focus: Rating Prediction Accuracy)*

These metrics are used when the recommender predicts **numerical ratings** (e.g., from 1 to 5 stars).

---

#### **2.2.1 Root Mean Squared Error (RMSE)**  
Measures the **square root of the average squared differences** between predicted ratings and actual ratings.

$$
\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (r_i - \hat{r}_i)^2}
$$

- **Sensitive to outliers** due to squaring errors.  
- Lower RMSE → Better predictive accuracy.

**Use Case:** Systems that predict **explicit ratings** (e.g., Netflix’s 5-star system).

---

#### **2.2.2 Mean Absolute Error (MAE)**  
Calculates the **average absolute difference** between predicted ratings and actual ratings.

$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |r_i - \hat{r}_i|
$$

- **Less sensitive to outliers** than RMSE.  
- Easier to interpret as it directly represents **average prediction error**.

---

#### 🔴 **2.3 Beyond Accuracy: User Experience Metrics** *(Focus: User Satisfaction & Diversity)*

While accuracy is important, a good recommender should also consider **user engagement, novelty, and diversity**.

---

#### **2.3.1 Coverage**  
**Coverage** measures the **proportion of items** in the catalogue that can be recommended.

$$
\text{Coverage} = \frac{\text{number of recommended items}}{\text{Total number of items}}
$$

- **Low coverage** → Only popular items are recommended.  
- **High coverage** → Niche/long-tail items are also promoted.

---

#### **2.3.2 Diversity**  
**Diversity** assesses the **variety** within the recommendation list. It helps avoid suggesting items that are too similar.

**Example metric:**  
- **Intra-list Distance** (average dissimilarity between recommended items).  
- **Jaccard Distance** or **Cosine Similarity** can be used to compute item dissimilarity.

$$
\text{Diversity} = \frac{\sum_{i \neq j} (1 - \text{sim}(i,j))}{\binom{n}{2}}
$$

---

#### **2.3.3 Novelty**  
**Novelty** measures how **new or unknown** the recommended items are to the user.  

**High novelty** → Users discover content they wouldn’t have found otherwise.  
- Can be measured by **item popularity** (recommending less popular items = higher novelty).

---

#### **2.3.4 Serendipity**  
**Serendipity** extends novelty by focusing on **pleasant surprises**. It suggests items that are both **unexpected** and **relevant**.

**Example:**  
- If a user loves sci-fi but is recommended an indie time-travel movie that they end up loving, that’s serendipitous.

---

### 📐 **3. Choosing the Right Metric: When to Use What?**

| **Metric**     | **Use Case**                           | **Focus**               |
|----------------|---------------------------------------|-------------------------|
| RMSE, MAE      | Rating prediction tasks               | Predictive accuracy     |
| Precision@K    | E-commerce, top-N recommendations     | Relevance               |
| Recall@K       | Medical, legal, critical domains      | Exhaustiveness          |
| MAP, NDCG      | Search engines, ranked lists          | Ranking quality         |
| Diversity      | Music, news, entertainment            | Exploration             |
| Novelty        | Social media, streaming platforms     | Content discovery       |
| Serendipity    | Book, film recommendations            | Unexpected relevance    |
| Coverage       | Niche markets, long-tail content      | Catalog exposure        |

#### **4.2 NDCG@K**

````python
def ndcg_at_k(r: list[int], k: int) -> float:
    """ndcg_at_k - Normalised Discounted Cumulative Gain at k

    Parameters
    ----------
    r : list[int]
        relevance scores (list or numpy) in rank order (first element is the first item)
    k : int
        number of results to consider

    Returns
    -------
    float
        ndcg@k metric value.
    """
    r = np.asfarray(r)[:k]  # Consider top-k results

    dcg = np.sum(r / np.log2(np.arange(2, r.size + 2)))  # DCG formula

    ideal_r = sorted(r, reverse=True)  # Ideal ranking
    idcg = np.sum(ideal_r / np.log2(np.arange(2, len(ideal_r) + 2)))  # IDCG formula

    # Normalise
    return dcg / idcg if idcg > 0 else 0


# Example relevance scores (binary or graded)
relevance_scores = [3, 2, 3, 0, 1, 2]
ndcg = ndcg_at_k(relevance_scores, 5)
print(f"NDCG@5: {ndcg:.2f}")
````

#### Understanding NDCG@K (Normalised Discounted Cumulative Gain)

As mentioned above, the Normalised Discounted Cumulative Gain (NDCG) is a popular ranking evaluation metric used in recommender systems, search engines, and other information retrieval systems. It not only checks whether relevant items are retrieved but also how highly they are ranked — rewarding systems that place the most relevant items at the top.

**What Does NDCG Measure?**
**Relevance**: It evaluates how relevant the recommended items are to the user.
**Ranking Quality**: It gives higher importance to items that appear earlier in the recommendation list.
**Normalisation**: By normalising against the ideal ranking, it ensures the final score is between 0 and 1, where:
* 1.0 → Perfect ranking (most relevant items at the top).
* 0.0 → Worst possible ranking.

**Why Does NDCG@5 Return 0.97 Here?**
Given:

- relevance_scores = [3, 2, 3, 0, 1, 2]
- We consider only the top 5 items: [3, 2, 3, 0, 1]

* The ideal ranking for these 5 items would be: [3, 3, 2, 1, 0]
* The DCG for the actual ranking is: $3/log(2) + 2/log(3) + 3/log(4) + 0/log(5) + 1/log(6) \approx 6.15$
* The DCG for the ideal ranking is: $3/log(2) + 3/log(3) + 2/log(4) + 1/log(5) + 0/log(6) \approx 6.32$
* The NDCG@5 is: $6.15/6.32 \approx 0.97$

The NDCG@5 is close to 1, indicating the ranking is almost perfect.
The slight drop from 1.0 is due to the second and third items being swapped compared to the ideal ranking:
The second most relevant item (relevance score 3) appears in the third position instead of the second.
Since this is a minor misplacement, the NDCG is only slightly below 1.

### Example: Comparison of two models trained to optimise different metrics.

We will compare two recommender models using different metrics. The imaginary case is the following, we take a user and two recommender models, we are going to analyse the recommendations to the same user and compare the results, without worrying for the moment about the training process of the models.
We also imagine we have a metric to compare two movies and we can calculate the similarity between movies.

#### Prepare Two Recommender Models
* Model A (Accuracy-Optimised)
    - Focuses purely on Precision, Recall, or NDCG.
    - Example: Recommender system trained to minimise RMSE or maximise Precision@K.

* Model B (Beyond Accuracy)
    - Adds diversity, novelty, and serendipity constraints.
    - Techniques to improve beyond-accuracy:
        - **Diversity** → Penalise similar items using dissimilarity matrices.
        - **Novelty** → Introduce long-tail or less popular items.
        - **Serendipity** → Recommend unexpected yet relevant items.

#### Define Metrics for Comparison

| Metric	| What It Evaluates | Calculation Example |
|---|---|---|
| **Precision@K**	| Relevance of recommendations	| `sklearn.metrics.precision_score` |
| **Recall@K**	| Ability to retrieve all relevant items	| `sklearn.metrics.recall_score` |
| **Diversity**	| Dissimilarity between recommended items	| `1 - Cosine Similarity` |
| **Novelty**	| Frequency/popularity of recommended items	| `Inverse Popularity` |
| **Serendipity**	| Relevance + Surprise (not obvious recommendations)	| `Similarity to user history` |

#### The code

Here is the implementation of a python code to evaluate the two different recommendation systems.


````python
# Sample recommendation lists for two models
user_recommendations_A = ["Movie1", "Movie2", "Movie3", "Movie4", "Movie5"]
user_recommendations_B = ["Indie1", "Movie2", "Docu1", "Movie5", "Classic1"]

# User's watch history
user_history = ["Movie2", "Movie5", "Classic2"]

# Movie popularity (lower popularity = higher novelty)
movie_popularity = {
    "Movie1": 0.9,
    "Movie2": 0.95,
    "Movie3": 0.85,
    "Movie4": 0.88,
    "Movie5": 0.9,
    "Indie1": 0.2,
    "Docu1": 0.3,
    "Classic1": 0.15,
    "Classic2": 0.1,
}

# Pairwise movie similarities (0 = dissimilar, 1 = identical)
movie_similarity = {
    ("Movie1", "Movie2"): 0.9,
    ("Movie1", "Indie1"): 0.2,
    ("Movie1", "Docu1"): 0.3,
    ("Movie2", "Classic1"): 0.4,
    ("Movie5", "Classic1"): 0.5,
}


# Calculate Diversity
def diversity(recommendations):
    similarities = []
    for i in range(len(recommendations)):
        for j in range(i + 1, len(recommendations)):
            pair = (recommendations[i], recommendations[j])
            sim = movie_similarity.get(pair, 0.1)  # Assume low similarity if unknown
            similarities.append(sim)
    return 1 - np.mean(similarities)


# Calculate Novelty
def novelty(recommendations):
    pop_scores = [movie_popularity.get(movie, 0.5) for movie in recommendations]
    return 1 - np.mean(pop_scores)  # Less popular = more novel


# Calculate Serendipity
def serendipity(recommendations, history):
    serendipity_scores = []
    for rec in recommendations:
        if rec not in history:
            sim = np.mean([movie_similarity.get((rec, h), 0.1) for h in history])
            serendipity_scores.append(1 - sim)  # More dissimilar = more serendipitous
    return np.mean(serendipity_scores)


# Compare Models
for model, recs in [
    ("Model A", user_recommendations_A),
    ("Model B", user_recommendations_B),
]:
    div = diversity(recs)
    nov = novelty(recs)
    ser = serendipity(recs, user_history)
    print(f"\n📊 {model} Results:")
    print(f"Diversity: {div:.2f}")
    print(f"Novelty: {nov:.2f}")
    print(f"Serendipity: {ser:.2f}")
````

**TF-IDF vectoriser**: We need to convert our text into numerical data. There are a huge number of ways to do so, one of the simplest is TF-IDF vectoriser.

To give a brief explanation of how it works, TF-IDF stands for *Term Frequency-Inverse Document Frequency*. It is a numerical statistic that reflects how important a word is to a specific document within a larger corpus. It balances two competing ideas:

- **Term Frequency (TF)**: How often a term appears in a document. The more times a word appears, the more relevant it might be — but this needs moderation to avoid overvaluing common terms.

- **Inverse Document Frequency (IDF)**: How rare the term is across all documents in the corpus. Words that appear in many documents (like "the", "is", "and") are less informative and therefore down-weighted.

The TF-IDF value increases proportionally to the number of times a word appears in the document but is often offset by the frequency of the word in the corpus, which helps to adjust for the fact that some words appear more frequently in general.

**Example**
A simple example of TF-IDF is shown below:
```
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]
```
then the _dictionary_ is the set of all unique words appearing in the corpus:

```
dictionary = {"this", "is", "the", "first", "document", "second", "and", "third", "one", "?"}
```

The general formula for TF-IDF is:
$$ \text{TF-IDF}(t, d, D) = \text{TF}(t, d) \times \text{IDF}(t, D) $$

where:
- $t$ is the term (word)
- $d$ is the document
- $D$ is the corpus (collection of documents)

Breaking Down the Components:

**TF (Term Frequency):**
The frequency of term $t$ in document $d$, typically normalised by the total number of terms in the document to avoid favouring longer documents.

$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$

**IDF (Inverse Document Frequency):**
Measures how important a term is across the entire corpus. It is the logarithm of the total number of documents divided by the number of documents containing the term, and is used to down-weight common terms.

$$ \text{IDF}(t, D) = \log \left( \frac{\text{Total number of documents}}{\text{Number of documents containing term } t} \right) $$

This means that the TF-IDF score for a term in a document is the product of the term frequency (TF) (how many times the word appears in a document, normalised by the document length, to not favorise longer documents) and the inverse document frequency (IDF) (how common or rare the word is across all documents, normalised by the number of documents, to not favorise rare words that appear in only one document).

#### 📊 **Calculating TF-IDF for the Word 'document' in the First Document**

Let's calculate the TF-IDF for the word 'document' in the first document in our toy case.

- Term Frequency (TF) for 'document' in Doc 1:

'document' appears $1$ time in Doc 1. Total words in Doc 1: $5$ ("this is the first document").
$$
\text{TF} = \frac{1}{5} = 0.2
$$

- Inverse Document Frequency (IDF) for 'document':
'document' appears in $3$ out of $4$ documents. Total documents = $4$.
$$
\text{IDF} = \log\left(\frac{4}{3+1}\right) = 0
$$

- TF-IDF for 'document' in Doc 1:
$$
\text{TF-IDF} = 0.2 \times 0 = 0
$$
This shows how TF-IDF naturally penalizes common terms.

#### 📊 **Calculating TF-IDF for the Word 'second' in the Second Document**

We want to calculate the **TF-IDF** score for the word **'second'** in the **second document** of the same corpus.

- Term Frequency (TF) for 'second':
It appears only once in the second document. The total number of words in the second document is $6$.
$$
\text{TF} = \frac{1}{6} = 0.1667
$$

- Inverse Document Frequency (IDF) for 'second':
The word 'second' appears in $1$ out of $4$ documents.
$$
\text{IDF} = \log\left(\frac{4}{1+1}\right) = \log(2) \approx 0.6931
$$
- TF-IDF for 'second' in the second document:
$$
\text{TF-IDF} = 0.1667 \times 0.6931 \approx 0.1160
$$

These scores can be used to associate a number to each word in each document, which can then be used to compare documents and find similarities between them.

This is the basic idea behind TF-IDF, a popular method for text analysis and information retrieval. At the end we get a matrix where each row represents a document and each column represents a word, with the cell values being the TF-IDF scores.

### KNN Recommender System

We can now use this logic to build our KNN recommenderder system. The system will recommend the top K documents based on the graded relevance scores. The KNN recommender system will be used to recommend documents to users based on their query. The query 
will be the user's input and the documents will be the recommended documents.

````python
def knn_recommender(query_vector, doc_vectors, top_n=5):
    """
    KNN Recommender to find top-N similar documents to the given query.

    Parameters
    ----------
    query_vector : ndarray
        The vectorized form of the query.
    doc_vectors : ndarray
        The vectorized form of the document corpus.
    top_n : int, optional
        The number of top similar documents to return, by default 5.

    Returns
    -------
    tuple
        distances : ndarray
            The distances (1 - cosine similarity) of the top-N documents.
        indices : ndarray
            The indices of the top-N similar documents.
    """
    knn = NearestNeighbors(n_neighbors=top_n, metric="cosine")
    knn.fit(doc_vectors)

    # Ensure query_vector is 2D
    if len(query_vector.shape) == 1:
        query_vector = query_vector.reshape(1, -1)

    distances, indices = knn.kneighbors(query_vector)
    return distances, indices


# ----------------------------------------------
# ✅ Sample Usage of KNN Recommender
# ----------------------------------------------

# Sample corpus
documents_example = [
    "Deep learning revolutionizes AI.",
    "Natural language processing models are complex.",
    "Astrophysics explores the universe.",
    "Bioinformatics integrates biology and data science.",
    "Nuclear energy is a powerful resource.",
    "Artificial intelligence applications are growing rapidly.",
    "Political sciences study social behavior.",
]

# Sample queries
queries_example = [
    "Deep Learning",
    "Science",
    "Artificial Intelligence",
]

# ✅ TF-IDF Vectoriser
vectoriser = TfidfVectorizer(
    lowercase=True,  # Convert to lowercase
    stop_words="english",  # Remove common stopwords
    ngram_range=(1, 2),  # Use unigrams and bigrams
    smooth_idf=True,
)

# Fit the vectorizer on the documents
doc_example_vectors = vectoriser.fit_transform(documents_example)

# Transform queries using the same fitted vectorizer
query_example_vectors = vectoriser.transform(queries_example)

# Example: Run KNN Recommender for a query
query_index = 1  # For "Science"
query_example_vec = query_example_vectors[query_index]

# Run KNN Recommender
distances, indices = knn_recommender(query_example_vec, doc_example_vectors, top_n=5)

# Convert distances to similarity scores
graded_relevance = 1 - distances[0]  # Cosine similarity = 1 - distance

# Display Top 5 Recommendations
print(f"\n🔍 Query: '{queries_example[query_index]}'")
print("Top 5 Relevant Documents with Graded Scores:")
for i, idx in enumerate(indices[0]):
    print(f"Doc {idx+1} (Relevance Score: {graded_relevance[i]:.4f})")
    print(f"Content: {documents_example[idx]}\n")
````

````python
# Assuming the following variables are already defined:
# - queries: list of query strings
# - query_vectors: vectorized representations of the queries
# - doc_vectors: vectorized representations of the documents
# - ground_truth_df: DataFrame containing ground truth relevance scores
# - knn_recommender: function that returns distances and indices for top-N documents


def precision_recall_at_k(graded_relevance, ground_truth, k, threshold=0.3):
    """Calculate Precision@K and Recall@K with Ground Truth."""
    relevance_at_k = graded_relevance[:k]

    true_positives = sum(
        1
        for idx, score in enumerate(relevance_at_k)
        if score >= threshold and ground_truth[idx] > 0
    )

    precision = true_positives / k
    total_relevant = sum(1 for score in ground_truth if score > 0)
    recall = true_positives / total_relevant if total_relevant > 0 else 0

    return precision, recall


def ndcg_at_k(graded_relevance, ground_truth, k):
    """Calculate NDCG@K with Ground Truth."""
    r = np.asfarray(graded_relevance)[:k]
    dcg = np.sum(r / np.log2(np.arange(2, r.size + 2)))

    ideal_r = sorted(ground_truth, reverse=True)[:k]
    idcg = np.sum(np.array(ideal_r) / np.log2(np.arange(2, len(ideal_r) + 2)))

    return dcg / idcg if idcg > 0 else 0


# ✅ 4. Evaluate Across Multiple Queries
results = []

for i, query in enumerate(queries_example):
    query_example_vec = query_example_vectors[i]
    distances, indices = knn_recommender(
        query_example_vec, doc_example_vectors, top_n=5
    )
    graded_relevance = 1 - distances[0]  # Convert distance to similarity

    # Extract ground truth relevance for the current query
    ground_truth_scores = ground_truth_df.iloc[indices[0], i].tolist()

    # Compute metrics with ground truth
    precision, recall = precision_recall_at_k(
        graded_relevance, ground_truth_scores, k=5
    )
    ndcg = ndcg_at_k(graded_relevance, ground_truth_scores, k=5)

    results.append(
        {"Query": query, "Precision@5": precision, "Recall@5": recall, "NDCG@5": ndcg}
    )

# Convert results to DataFrame
results_df = pd.DataFrame(results)
print(results_df)

# ✅ 5. Visualize the Evaluation Results

# Plot Precision@5
plt.figure(figsize=(12, 6))
plt.barh(results_df["Query"], results_df["Precision@5"], color="skyblue")
plt.xlabel("Precision@5")
plt.title("Precision@5 Across Queries")
plt.show()

# Plot Recall@5
plt.figure(figsize=(12, 6))
plt.barh(results_df["Query"], results_df["Recall@5"], color="salmon")
plt.xlabel("Recall@5")
plt.title("Recall@5 Across Queries")
plt.show()

# Plot NDCG@5
plt.figure(figsize=(12, 6))
plt.barh(results_df["Query"], results_df["NDCG@5"], color="lightgreen")
plt.xlabel("NDCG@5")
plt.title("NDCG@5 Across Queries")
plt.show()
````
