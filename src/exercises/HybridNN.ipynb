# Exercise: Hybrid Movie Recommender System

## Objective
Create a hybrid recommender system that combines both content-based filtering (using neural networks) and collaborative filtering approaches to provide movie recommendations.

## Data
The data to use is the notorious [disease symptom description dataset](https://www.kaggle.com/datasets/itachi9604/disease-symptom-description-dataset/data) available on kaggle.

Before starting, try to think about the correspondence between disease, symptoms and users, items. How would you define users, items, and interactions in this context?

## Tasks
### 1. Data Preparation and Feature Engineering
- Use the MovieLens dataset provided in the lecture
- Create additional features for movies beyond the existing ones:
- Extract year from release date as a separate feature
- Create genre interaction features (e.g., Action-SciFi, Drama-Romance)
- Add popularity metrics (number of ratings, average rating)
### 2. Neural Network Architecture Design
- Design a neural network architecture that:
- Takes both movie content features and user features as input
- Has at least 3 hidden layers with different activation functions
- Uses dropout layers to prevent overfitting
- Implements batch normalization
- Outputs a predicted rating
### 3. Training and Evaluation
- Split the data into training, validation, and test sets
- Implement a custom training loop that:
- Uses mini-batch gradient descent
- Implements early stopping
- Tracks both training and validation loss
- Evaluate the model using multiple metrics:
- Mean Squared Error (MSE)
- Mean Absolute Error (MAE)
- Precision@K and Recall@K
### 4. Hybrid Recommendation System
- Combine the neural network predictions with collaborative filtering:
- Use the neural network for content-based predictions
- Implement a simple collaborative filtering component
- Create a weighted combination of both approaches
- Experiment with different weighting schemes and evaluate their impact
### 5. Visualization and Analysis
- Create visualizations showing:
- Training/validation loss curves
- Feature importance analysis
- Comparison of pure content-based vs. hybrid approach
- Sample recommendations with explanations

## Deliverables
- A Jupyter notebook with:
    * Complete implementation
    * Visualizations
    * Performance analysis
    * Discussion of results
- A brief report (max 2 pages) explaining:
    * Design decisions
    * Challenges faced
    * Performance comparison
    * Potential improvements

## Bonus Challenge
Implement attention mechanisms in the neural network to better capture the importance of different features in the recommendation process.
